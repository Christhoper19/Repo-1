Is there a significant difference in median value of houses bounded by the Charles river or not? (T-test for independent samples)

  α = 0.05

```python
scipy.stats.levene(ratings_df[ratings_df["CHAS"]== 1]["MEDV"],
                      ratings_df[ratings_df["CHAS"]== 0]["MEDV"], center = "mean")
```

```python
scipy.stats.ttest_ind(ratings_df[ratings_df["CHAS"]== 1]["MEDV"],
                      ratings_df[ratings_df["CHAS"]== 0]["MEDV"], equal_var = False)

```



*   **Conclusion: based on the results of the independent samples T test, the p value obtained was 0.003, which is lower than the significance level of 0.05. Therefore, the null hypothesis is reject, indicating that there is statiscally significant difference in the median value of house between those bounded by the Charles River and those thar are not.**



Is there a difference in Median values of houses (MEDV) for each proportion of owner occupied units built prior to 1940 (AGE)? (ANOVA)

```python
a =  ratings_df[ratings_df["age group"] == "35 years and younger"]["MEDV"]
b = ratings_df[ratings_df["age group"] == "between 35 and 70"]["MEDV"]
c = ratings_df[ratings_df["age group"] == "70 years and older"]["MEDV"]
```

```python
f_statistic, p_value = scipy.stats.f_oneway(a, b, c )
print("F_statistic: {0}, P-Value: {1}".format(f_statistic, p_value))
```



*   Conclusion: Based on the One-Way ANOVA test, the F-statistic value is 36,41 with a p value of 1.71 x 10⁻¹⁵, which is less than the significance level of 0.05. Therefore, the null hypotesis is rejected, indicating that there is statistically significant difference in the median value of houses (MEDV) among the three age group of owner occupied units.

\What is the impact of an additional weighted distance  to the five Boston employment centres on the median value of owner occupied homes? (Regression analysis)

```python
X = ratings_df["DIS"]
y = ratings_df["MEDV"]

X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
model.summary()
```



**Conclusion:** The regression analysis shows that the variable weighted distance to five boston employment centre (DIS) has a positive and statistically significant effect on th emedian value of owner occupied homes (MEDV).

The estimate regression coefficient of 1.0916
indicates that for each additional unit of distance (DIS) , the median home value increases by aprroximately 1.09 thousand dollar.

the associated p value (1.21 × 10⁻⁸) is below the significance level of 0.05, confirming that the relationhsip is statistically significant.
